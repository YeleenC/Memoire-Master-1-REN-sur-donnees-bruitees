```python
import spacy
from spacy.tokens import DocBin
import json
import os

# Charger un modèle spaCy vide
nlp = spacy.blank("fr")

# Dossier contenant les fichiers JSON
json_folder = "JSON_entrainementC1"  # Chemin vers ton dossier local

# Fonction pour créer un .spacy à partir du JSON
def convert_json_to_spacy(json_path, output_path):
    with open(json_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    if not isinstance(data, list):  # On s'assure que c'est une liste d'objets
        data = [data]

    doc_bin = DocBin()  # Conteneur spaCy
    for item in data:
        text = item["content"]
        doc = nlp(text)

        ents = []
        for ann in item["annotations"]:
            start = ann["start_offset"]
            end = ann["end_offset"]
            label = ann["tag"]["name"]
            span = doc.char_span(start, end, label=label)
            if span is not None:
                ents.append(span)
        doc.ents = ents
        doc_bin.add(doc)

    doc_bin.to_disk(output_path)
    print(f"Fichier .spacy créé : {output_path}")

# Parcourir tous les fichiers JSON dans le dossier
for json_name in os.listdir(json_folder):
    if json_name.endswith(".json"):  # Vérifier que c'est un fichier JSON
        json_path = os.path.join(json_folder, json_name)
        output_spacy = os.path.join(json_folder, json_name.replace(".json", ".spacy"))
        convert_json_to_spacy(json_path, output_spacy)
```
