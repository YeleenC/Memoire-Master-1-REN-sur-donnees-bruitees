```python
import spacy
from pathlib import Path
import json
import os
```

```python
MODELE_PATH = "/content/drive/MyDrive/Colab Notebooks/Memoire_M1/Entrainement/Modeles_enrichi_quaero/Tesseract_enrichi_quaero/model-best"
```

```python
TEXTES_PATH = "/content/drive/MyDrive/Colab Notebooks/Memoire_M1/Evaluation_DARIEN/DARIEN"
```

```python
SORTIE_PATH = "/content/drive/MyDrive/Colab Notebooks/Memoire_M1/Evaluation_DARIEN/resultats_Tesseract_enrichi_quaero"
os.makedirs(SORTIE_PATH, exist_ok=True)
```

```python
# Liste des versions du texte à traiter (référence, Tesseract et Kraken ici)
FICHIERS_TEST = [
    ("TESSERACT", "DARIEN_Le-voleur_tesseract0.3.10.txt"),
    ("REF", "DARIEN_Le-voleur_REF.txt"),
    ("KRAKEN", "DARIEN_Le-voleur_kraken4.3.13.dev25.txt")
]
```

```python
nlp = spacy.load(MODELE_PATH)
for dossier, nom_fichier in FICHIERS_TEST:
    chemin_texte = Path(TEXTES_PATH) / dossier / nom_fichier
    chemin_sortie = Path(SORTIE_PATH) / nom_fichier.replace(".txt", "_tok2vec_Tesseract_enrichi_quaero.json")

    with open(chemin_texte, "r", encoding="utf-8") as f:
        texte = f.read()

    doc = nlp(texte)

    resultat = {}
    compteur = 0

    for ent in doc.ents:
        resultat[f"entite_{compteur}"] = {
            "label": ent.label_,
            "text": ent.text,
            "jalons": [ent.start_char, ent.end_char]
        }
        compteur += 1

    with open(chemin_sortie, "w", encoding="utf-8") as f_out:
        json.dump(resultat, f_out, indent=2, ensure_ascii=False)

    print(f"Fichier traité : {nom_fichier} | Entités détectées : {compteur}")
```
